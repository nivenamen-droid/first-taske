```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
import pandas as pd

file_name = 'weather _data.xls'

# Using pd.read_excel as the file has a .xls extension
try:
    data = pd.read_excel(file_name)
    data = data.dropna() # Drop rows with missing values
    x = data[['Humidity_pct','Precipitation_mm','Wind_Speed_kmh']]
    y = data['Temperature_C']
    print("Data loaded and 'x', 'y' defined successfully using pd.read_excel.")
except FileNotFoundError:
    print(f"Error: The file '{file_name}' was not found. Please ensure it's uploaded to your environment.")
    x = pd.DataFrame() # Initialize to prevent NameError in subsequent cells
    y = pd.Series()
except Exception as e:
    print(f"An error occurred while reading the file: {e}")
    print("Ensure the file is a valid Excel file and the format is correct.")
    x = pd.DataFrame() # Initialize to prevent NameError in subsequent cells
    y = pd.Series()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
```

```python
print(f'Mean Squared Error: {mse}')
print(f'R-squared: {r2 * 100}')
```

```python
import matplotlib.pyplot as plt
```

```python
plt.bar(x_train.columns,model.coef_)
plt.xlabel('Features')
plt.ylabel('Coefficients')
plt.title('Feature Coefficients')
plt.show()
```
